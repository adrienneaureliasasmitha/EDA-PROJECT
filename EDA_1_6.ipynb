```python
import numpy as np

harga = np.array([5000, 7000, 3000, 12000, 4500])

print('Rata-rata harga:', harga.mean())
print('Harga tertinggi:', harga.max())
print('Harga setelah diskon 10%:', harga * 0.9)
```

NumPy Array dapat melakukan operasi matematika secara langsung terhadap seluruh elemen array. Ketika menggunakan harga * 0.9, setiap harga otomatis dikalikan dengan 0,9.

Berbeda dengan list Python biasa, perintah [5000, 7000, 3000] * 0.9 akan menghasilkan error karena list tidak mendukung perkalian dengan bilangan desimal. Oleh karena itu, NumPy lebih praktis untuk melakukan perhitungan numerik pada banyak data sekaligus.

```python
import pandas as pd

df = pd.read_csv('dataset_penjualan_kantin (2).csv')

print("5 BARIS PERTAMA")
print(df.head())

print("\nINFORMASI DATA")
df.info()

print("\nSTATISTIK DATA")
print(df.describe())

print("\nUKURAN DATA")
print(df.shape)
```

Berdasarkan hasil df.info(), terdapat beberapa kolom yang jumlah data non-null-nya lebih sedikit dibandingkan jumlah keseluruhan baris, yaitu:
- jumlah_terjual: 65 data.
- harga_satuan: 66 data.
- nama_kasir: 66 data.
Dataset memiliki 69 baris, sehingga terdapat data kosong pada ketiga kolom tersebut. Hal ini menunjukkan bahwa dataset masih memerlukan proses pembersihan sebelum dianalisis lebih lanjut.

```python
# Mengecek jumlah data kosong
print("Jumlah Missing Value:")
print(df.isnull().sum())

# Mengubah kolom jumlah_terjual menjadi numerik
df['jumlah_terjual'] = pd.to_numeric(
    df['jumlah_terjual'].astype(str).str.extract(r'(\d+)')[0],
    errors='coerce'
)

# Mengubah harga menjadi numerik
df['harga_satuan'] = pd.to_numeric(
    df['harga_satuan'].astype(str).str.replace(
        r'[^0-9]', '', regex=True
    ),
    errors='coerce'
)

# Mengisi data kosong jumlah terjual dengan 0
df['jumlah_terjual'] = df['jumlah_terjual'].fillna(0)

# Mengisi harga kosong dengan median harga
df['harga_satuan'] = df['harga_satuan'].fillna(
    df['harga_satuan'].median()
)

# Mengisi nama kasir yang kosong
df['nama_kasir'] = df['nama_kasir'].fillna('Tidak diketahui')

# Mengecek kembali missing value
print("\nMissing Value Setelah Cleaning:")
print(df.isnull().sum())
```

Kolom jumlah_terjual diisi dengan nilai 0 sesuai asumsi bahwa transaksi tanpa informasi jumlah dianggap tidak memiliki jumlah penjualan yang tercatat. Sementara itu, harga_satuan diisi menggunakan nilai median agar data kosong dapat dilengkapi tanpa terlalu terpengaruh oleh nilai ekstrem.

Untuk nama_kasir, data kosong diganti menjadi Tidak diketahui karena identitas kasir tidak dapat ditentukan dari data yang tersedia. Dengan demikian, informasi transaksi tetap dapat dipertahankan.

Catatan: pengisian jumlah terjual dengan 0 merupakan asumsi pengolahan, bukan bukti bahwa barang benar-benar tidak terjual.

```python
# Mengecek jumlah data duplikat
print("Jumlah Data Duplikat:")
print(df.duplicated().sum())

# Menghapus data duplikat
df = df.drop_duplicates()

# Memastikan tipe data sesuai
df['jumlah_terjual'] = df['jumlah_terjual'].astype(int)
df['harga_satuan'] = df['harga_satuan'].astype(int)

# Menampilkan tipe data
print("\nTipe Data Setelah Cleaning:")
print(df.dtypes)

# Menampilkan jumlah baris setelah penghapusan duplikat
print("\nJumlah Baris Setelah Cleaning:")
print(df.shape)
```

Berdasarkan pemeriksaan dataset, ditemukan 4 baris duplikat dari total 69 baris. Setelah menggunakan drop_duplicates(), jumlah data menjadi 65 baris.

Pemeriksaan tipe data penting dilakukan agar setiap kolom dapat digunakan sesuai fungsinya. Misalnya, jumlah_terjual dan harga_satuan harus menggunakan tipe data numerik agar dapat digunakan dalam operasi matematika dan perhitungan pendapatan.

```python
# 1. Filtering
# Menampilkan transaksi dengan jumlah penjualan lebih dari 10
laris = df[df['jumlah_terjual'] > 10]

print("DATA FILTERING:")
print(laris)

# 2. Sorting
# Mengurutkan transaksi berdasarkan jumlah penjualan tertinggi
urut = df.sort_values(
    by='jumlah_terjual',
    ascending=False
)

print("\nDATA SORTING:")
print(urut)

# 3. Membuat kolom turunan
df['total_pendapatan'] = (
    df['harga_satuan'] * df['jumlah_terjual']
)

print("\nDATA DENGAN KOLOM TOTAL PENDAPATAN:")
print(df.head())

# 4. Groupby dan agregasi
ringkasan = df.groupby('nama_produk')[
    'total_pendapatan'
].sum().sort_values(ascending=False)

print("\nRINGKASAN PENDAPATAN SETIAP PRODUK:")
print(ringkasan)
```

Berdasarkan hasil pengelompokan menggunakan groupby(), produk dengan total pendapatan tertinggi adalah Roti Bakar. Informasi ini dapat membantu pihak kantin dalam mengetahui produk yang memberikan kontribusi pendapatan terbesar, sehingga dapat menjadi pertimbangan dalam mengatur persediaan dan strategi penjualan.

Namun, hasil tersebut perlu diperiksa lebih lanjut karena terdapat transaksi Roti Bakar dengan jumlah penjualan 500 unit yang jauh lebih tinggi dibandingkan transaksi lainnya.
